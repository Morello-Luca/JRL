Class breakdown 

```python
class Pokemon:
    def __init__(self,name,hp,defence,attack,level,element,strong,weak):
        self.name = name
        self.hp = hp
        self.defence = defence
        self.attck = attack
        self.level = level
        self.exp = 0
        self.element = element
        self.strong = strong
        self.weak = weak

    def level_up(self):
        self.level += 1
        self.exp = 0
        print(f"{self.name} leveled up to Level {self.level}!")

    def attack_target(self, target):
        """Calculates and deals damage to another Pokémon."""
        damage = max(1, self.attack - target.defence)
        print(f"{self.name} attacks {target.name} for {damage} damage!")
        target.take_damage(damage)

    def take_damage(self, amount):
        """Reduces HP and checks if the Pokémon has fainted."""
        self.hp -= amount

        if self.hp < 0:
            self.hp = 0
        print(f"-> {self.name} now has {self.hp}/{self.max_hp} HP.")

        if self.hp == 0:
            print(f"💀 {self.name} has fainted!")
    def is_fainted(self):
        """Returns True if HP is 0, else False."""
        return self.hp == 0
```

## Subclasses

```python
class ElectricPokemon(Pokemon):
    def __init__(self, name, hp, defence, attack, level):
        super().__init__(
            name=name, 
            hp=hp, 
            defence=defence, 
            attack=attack, 
            level=level, 
            element="Electric", 
            strong=["Water", "Flying"], 
            weak=["Ground"]
        )

    def thunderbolt(self, target):
        print(f"{self.name} uses Thunderbolt!")
        damage = max(1, (self.attack * 1.5) - target.defence)
        target.take_damage(int(damage))

class WaterPokemon(Pokemon):
    def __init__(self, name, hp, defence, attack, level):
        super().__init__(
            name=name, 
            hp=hp, 
            defence=defence, 
            attack=attack, 
            level=level, 
            element="Water", 
            strong=["Fire", "Ground", "Rock"], 
            weak=["Electric", "Grass"]
        )

    # Any water pokemon can use this!
    def surf(self, target):
            print(f"{self.name} uses Surf!")
            damage = max(1, (self.attack * 1.4) - target.defence)
            target.take_damage(int(damage))

```

Instances

```python
# --- Let's Simulate a Mini Battle ---
pikachu = ElectricPokemon(name="Pikachu", hp=100, defence=30, attack=50, level=10)
squirtle = WaterPokemon(name="Squirtle", hp=120, defence=45, attack=40, level=10)

print("--- BATTLE START ---")
# Pikachu attacks Squirtle with a normal attack
pikachu.attack_target(squirtle)

print("\n--- NEXT TURN ---")
# Squirtle fights back using its special water move
squirtle.surf(pikachu)
```
